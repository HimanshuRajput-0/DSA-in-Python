Question : What is searching in Data Structure ?

Ans = it is a process of locating specific element in the data structure based on given keys and value



*   Returns the position of the element in the data structure if it is found.
*   Different searching algorithms are used depending on the different data structure.



Question : What are different Searching algorithms ?

Ans = the choice of algorithm whether the data is sorted and the type of data structure.


*   Linear Search : checks each element until the target is found.
*   Binary Search : repeatedly divides the array into half until the target is found.

*   Jumps Search : Skips fixed sized blocks in the sorted array before performing a linear search.
*   Interpolated Search : stimates the target's position in a sorted, uninformly distributed array.


*   Exponential Search : finds a search range by finding the index, then applies binary search.
*   Hash Based Search : use a hash table to provide near constant-time lookup.
