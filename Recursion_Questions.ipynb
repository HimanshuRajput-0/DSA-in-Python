**Mathematical Questions:**

```
Given a number n, print the following pattern without using any loop.

n, n-5, n-10, ..., 0, 5, 10, ..., n+5, n
```

```Solution:```

````python
def pattern(n):
  if (n == 0 or n < 0):
    print(n, end=", ")
    return
  print(n, end=", ")
  pattern(n - 5)
  print(n, end=", ")

if __name__ == "__main__":
  n = int(input())

  pattern(n)
````
