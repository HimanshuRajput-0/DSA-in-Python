Data Structure are specialized objects for organizing data efficiently.

There are some data structures takes a long time to build, but once built their data are quickly accessible. Some built quickly, but not as efficiently accessible.

Python has several built in data structure classes, namely list,set, dict, and tuple.

# Linked list :

it is a data structure that mimics the built in list class, but has a different underlying implementation.

There is a node like a standard size box that can hold all the different types of objects.

Data in the node is stored as an attribute.

A linked list chains nodes together. Every linked list needs a reference to the first node in the chain, called the head.

A reference to the last node called the tail.

Each node instance in the list stores a piece of data, plus at least one reference to another node in the list.
